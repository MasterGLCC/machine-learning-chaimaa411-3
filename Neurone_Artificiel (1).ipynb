```python
import numpy as np   # Bibliothèque pour les calculs mathématiques et les matrices

# 1- Données d'entraînement
X = np.array([[0,0],[0,1]], dtype=float)  # Entrées
Y = np.array([[0],[1]], dtype=float)   # Sorties attendues

# 2- Initialisation des poids et du biais
np.random.seed(42)   # Fixe le hasard pour reproduire les mêmes résultats

W = np.random.randn(2, 1)  # 2 poids du neurone (un pour chaque entrée)
b = np.zeros((1, 1))       # biais (valeur ajoutée à la somme pondérée)

# 3- Fonction d'activation Sigmoïde : transforme une valeur en probabilité entre 0 et 1
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 4- Forward Propagation : Calcule la sortie du neurone à partir des entrées
def forward(X):
    z = X @ W + b          # Calcul de la somme pondérée : XW + b
    a = sigmoid(z)         # Application de la fonction d'activation
    return z, a            # z = valeur brute, a = prédiction finale

# 5- Backward Propagation : Met à jour les poids pour réduire l'erreur
def backward(X, Y, a, lr=0.1):
    global W, b       # Permet de modifier les variables globales
    m = X.shape[0]    # nombre d'exemples d'entraînement
    dz = a - Y        # Erreur entre prédiction et vérité

    # Calcul des gradients
    dW = (X.T @ dz) / m
    db = np.mean(dz, axis=0, keepdims=True)

    # Mise à jour des paramètres (descente de gradient)
    W -= lr * dW
    b -= lr * db

# 6- Fonction Logloss : Mesure l'erreur entre prédictions et vraies valeurs
def loss(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-8, 1 - 1e-8)    # Évite log(0) (erreur mathématique)
    
    return -np.mean(      # Formule de Log Loss
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )

# 7- Entraînement du neurone
for epoch in range(1001):
    z, a = forward(X)             # Forward : calcul des prédictions
    backward(X, Y, a, lr=0.5)     # Backward : mise à jour des poids

# 8- Calcul des prédictions après entraînement
_, predictions = forward(X)
for x, y, p in zip(X, Y, predictions):

    # Conversion probabilité → classe (0 ou 1)
    label = 1 if p[0] > 0.5 else 0
    print(
        f"Entrée {x.tolist()} "
        f"→ Prédit: {p[0]:.3f} "
        f"→ Classe: {label}"
    )
```
